```python
# Moving averages (trend indicator)
data["MA_20"] = data["Close"].rolling(20).mean()
data["MA_50"] = data["Close"].rolling(50).mean()

plt.figure(figsize=(10,5))
plt.plot(data["Close"], label="Close Price", alpha=0.5)
plt.plot(data["MA_20"], label="20-day MA")
plt.plot(data["MA_50"], label="50-day MA")
plt.title("AAPL Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.show()

# Volume spikes (crowd behavior / FOMO indicator)
data["Volume_MA_20"] = data["Volume"].rolling(20).mean()
data["Volume_Spike"] = data["Volume"] > (2 * data["Volume_MA_20"])

print("Number of volume spike days:", data["Volume_Spike"].sum())
```
